```python
# Collection 1: Thông tin cá nhân
personal_info = df[["customerID", "gender", "SeniorCitizen", "Partner", "Dependents"]]
db.personal_info.insert_many(personal_info.to_dict("records"))
```

```python
# Collection 2: Dịch vụ khách hàng đang sử dụng
services = df[["customerID", "PhoneService", "MultipleLines", "InternetService", "OnlineSecurity", 
               "OnlineBackup", "DeviceProtection", "TechSupport", "StreamingTV", "StreamingMovies"]]
db.services.insert_many(services.to_dict("records"))
```

```python
# Collection 4: Lịch sử khách hàng
customer_history = df[["customerID", "tenure", "Churn"]]
db.customer_history.insert_many(customer_history.to_dict("records"))
```
